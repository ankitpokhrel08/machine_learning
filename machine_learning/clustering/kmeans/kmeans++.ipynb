```python
#Creating a toy dataset with 5 Clusters with make_blobs
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
# Parameters
n_samples = 300
n_features = 2
centers = 5
# Generate data
x, y = make_blobs(n_samples=n_samples, n_features=n_features, centers=centers, cluster_std=1.5)
# * Visualize the data
plt.scatter(x[:, 0], x[:, 1], s=50)
plt.title("Generated Dataset for K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
#plotting all 10 instializations to see the difference along withe silhouette score
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans
fig, axes = plt.subplots(2, 5, figsize=(20, 8))
for i, ax in enumerate(axes.flatten()):
    kmeans = KMeans(n_init=i+1, init='random',n_clusters=5)
    kmeans.fit(x)
    y_kmeans = kmeans.predict(x)
    ax.scatter(x[:, 0], x[:, 1], c=y_kmeans, s=50, cmap='viridis')
    centers = kmeans.cluster_centers_
    ax.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75, marker='X')
    ax.set_title(f"(init {i+1}) with silhouette score: {silhouette_score(x, y_kmeans):.4f}")
    ax.set_xlabel("Feature 1")
    ax.set_ylabel("Feature 2")
plt.tight_layout()
plt.show()
```
