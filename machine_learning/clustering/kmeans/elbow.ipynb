```python
#Importing Kmeans
from sklearn.cluster import KMeans
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
```

```python
#Creating a toy dataset with 5 Clusters with make_blobs
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
# Parameters
n_samples = 300
n_features = 2
centers = 4
# Generate data
x, y = make_blobs(n_samples=n_samples, n_features=n_features, centers=centers, cluster_std=1.5, random_state=42)
# * Visualize the data
plt.scatter(x[:, 0], x[:, 1], s=50)
plt.title("Generated Dataset for K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
#Let's build KMeans for 1,2,3,4,5,6,7,8 clusters and calculate inertia for each model
inertia = []
K = range(1, 9)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(x)
    inertia.append(kmeans.inertia_)
```

```python
#Plottiung the inertia values
plt.figure(figsize=(8, 5))
plt.plot(K, inertia, 'bo-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method For Optimal k')
plt.xticks(K)
plt.show()
```

```python
#Plotting number of clusters vs selhouette score
from sklearn.metrics import silhouette_score
silhouette_scores = []
for k in K[1:]:  # Silhouette score is not defined for k=1
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(x)
    score = silhouette_score(x, kmeans.labels_)
    silhouette_scores.append(score)
plt.figure(figsize=(8, 5))
plt.plot(K[1:], silhouette_scores, 'bo-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score For Optimal k')
plt.xticks(K[1:])
plt.show()
```

```python
#Let's use n_clusters = 4 based on the elbow method
kmeans = KMeans(n_clusters=4)
kmeans.fit(x)
#Plotting the clusters
y_kmeans = kmeans.predict(x)
plt.scatter(x[:, 0], x[:, 1], c=y_kmeans, s=50, cmap='viridis')
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75, marker='X')
plt.title("K-Means Clustering with 4 Clusters")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
#Let's use n_clusters = 4 based on the elbow method
kmeans = KMeans(n_clusters=3)
kmeans.fit(x)
#Plotting the clusters
y_kmeans = kmeans.predict(x)
plt.scatter(x[:, 0], x[:, 1], c=y_kmeans, s=50, cmap='viridis')
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75, marker='X')
plt.title("K-Means Clustering with 4 Clusters")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
#Calculating the silhouette score
from sklearn.metrics import silhouette_score
score = silhouette_score(x, y_kmeans)
print(f'Silhouette Score for k=4: {score:.4f}')
```

```python
X, y = make_moons(n_samples=1000, noise=0.1, random_state=42)
```

```python
#Plotting the moons dataset
import matplotlib.pyplot as plt
plt.scatter(X[:, 0], X[:, 1])
plt.title("Moons Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
```

```python
#Using KMeans to cluster the data
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(X)
```

```python
#Plotting the clustered data with centroids
y_kmeans = kmeans.predict(X)
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap='viridis')
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75, marker='X')
plt.title("K-Means Clustering on Moons Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
#Calculating the silhouette score
from sklearn.metrics import silhouette_score
score = silhouette_score(X, y_kmeans)
print(f'Silhouette Score for k=4: {score:.4f}')
```

```python
#Using KMeans to cluster the data
kmeans = KMeans()
kmeans.fit(X)
```

```python
#Plotting the clustered data with centroids
y_kmeans = kmeans.predict(X)
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap='viridis')
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75, marker='X')
plt.title("K-Means Clustering on Moons Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
#Calculating the silhouette score
from sklearn.metrics import silhouette_score
score = silhouette_score(X, y_kmeans)
print(f'Silhouette Score for k=4: {score:.4f}')
```

```python
#Let's cluster for K = 2,3,4,5,6,7,8 and plot graph between k and silhouette score
silhouette_scores = []
K = range(2, 9)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    score = silhouette_score(X, kmeans.labels_)
    silhouette_scores.append(score)

#Plotting number of clusters vs selhouette score
plt.figure(figsize=(8, 5))
plt.plot(K, silhouette_scores, 'bo-')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score For Optimal k on Moons Dataset')
plt.xticks(K)
plt.show()
```
