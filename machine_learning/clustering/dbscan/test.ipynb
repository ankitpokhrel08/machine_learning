```python
#Creating a toy dataset with 5 Clusters with make_blobs
from sklearn.datasets import make_circles
import matplotlib.pyplot as plt
X, y = make_circles(n_samples=300, factor=0.3, noise=0.1)
plt.scatter(X[:, 0], X[:, 1])
plt.show()
```

```python
#Using Kmeans to cluster the data
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=2)
kmeans.fit(X)
#Plotting the clustered data with centroids
y_kmeans = kmeans.predict(X)
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap='viridis')
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75, marker='X')
plt.show()
```

```python
#Using DBSCAN to cluster the data
from sklearn.cluster import DBSCAN
dbscan = DBSCAN(eps=0.2, min_samples=2)
y_dbscan = dbscan.fit_predict(X)
plt.scatter(X[:, 0], X[:, 1], c=y_dbscan, s=50, cmap='viridis')
plt.show()
```
