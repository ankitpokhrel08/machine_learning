```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
```

```python
X,y= load_iris(return_X_y=True,as_frame=True)
```

```python
X
```

```python
y
```

```python
import scipy.cluster. hierarchy as shc
plt.figure(figsize=(10, 7))
plt.title("Iris Dendogram")
dend = shc.dendrogram(shc.linkage(X[['sepal length (cm)','sepal width (cm)']]))
```

```python
import scipy.cluster. hierarchy as shc
plt.figure(figsize=(10, 7))
plt.title("Iris Dendogram")
dend = shc.dendrogram(shc.linkage(X[['petal length (cm)','petal width (cm)']]))
```

```python
plt.scatter(X[['petal length (cm)']],X[['petal width (cm)']])
```

```python
plt.scatter(X[['sepal length (cm)']],X[['sepal width (cm)']])
```

```python
from sklearn.cluster import AgglomerativeClustering

cluster = AgglomerativeClustering(n_clusters=3)
labels_=cluster.fit_predict(X[['petal length (cm)','petal width (cm)']])
```

```python
labels_
```

```python
n = 0
for i in labels_:
    if i == 1:
        labels_[n] = 0
    elif i == 0:
        labels_[n] = 2
    else:
        labels_[n] = 1
    n = n +1
```

```python
labels_
```

```python
y
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y,labels_)
```

```python
plt.scatter(X[['petal length (cm)']],X[['petal width (cm)']],c=cluster.labels_)
plt.title('Cluster with Hierarchical Clustering')
plt.show()
```

```python
plt.scatter(X[['petal length (cm)']],X[['petal width (cm)']],c=y)
plt.title("The real cluster plotting!")
plt.show()
```

```python
plt.scatter(X[['sepal length (cm)']],X[['sepal width (cm)']],c=cluster.labels_)
plt.title('Cluster with Hierarchical Clustering')
plt.show()
```

### Type of Linkage

```python
from sklearn.cluster import AgglomerativeClustering

cluster = AgglomerativeClustering(n_clusters=3,linkage='complete')
labels_=cluster.fit_predict(X[['petal length (cm)','petal width (cm)']])
```

```python
n = 0
for i in labels_:
    if i == 1:
        labels_[n] = 1
    elif i == 0:
        labels_[n] = 2
    else:
        labels_[n] = 0
    n = n +1
```

```python
plt.scatter(X[['petal length (cm)']],X[['petal width (cm)']],c=cluster.labels_)
plt.title('Cluster with Hierarchical Clustering')
plt.show()
accuracy_score(y,labels_)
```
